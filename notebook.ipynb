## ML

```python
# x,y
sales_inv["Month_Num"]= sales_inv["Month"].dt.month
sales_inv["Year"]= sales_inv["Month"].dt.year
#sort the data 
sales_inv = sales_inv.sort_values(["SKU", "Month"]) #ensure stock is consistance for each month
sales_inv.head(10)
```

```python
#feature engeneering
sales_inv["Prev_sales"]=(
    sales_inv.groupby("SKU")["Units Sold"].shift(1) #first month nan
sales_inv[["SKU", "Month", "Units Sold", "Prev_sales"]].head(10)
```

```python
sales_inv.columns
```

```python
#choose x y
num_cols=['Month_Num','Year','Prev_sales']
cat_cols=['SKU','Category'] 
```

```python
new_df = sales_inv.dropna(subset=['Prev_sales']).copy()
```

```python
new_df.shape #removes the ones with 0stock
```

```python
from sklearn.model_selection import train_test_split
new_df = new_df.sort_values('Month')
x = new_df[num_cols + cat_cols]
y = new_df['Units Sold']

#split1
x_train_val, x_test, y_train_val, y_test= train_test_split(x,y, test_size=0.2,random_state=42)
#split2 the train to train and val 
x_train, x_val, y_train, y_val = train_test_split(x_train_val,y_train_val, test_size=0.2,random_state=42)
```

```python
#pipline
from sklearn.pipeline import make_pipeline #seq 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, RobustScaler, FunctionTransformer, StandardScaler
from sklearn.compose import ColumnTransformer

num_pipe= make_pipeline(SimpleImputer(strategy='median'),StandardScaler())
cat_pipe = make_pipeline(OneHotEncoder(handle_unknown='ignore',sparse_output=False))
```

```python
#transform
from sklearn.compose import ColumnTransformer
preprocessing_pipe= ColumnTransformer(
    [
        ('num', num_pipe,num_cols ),
        ('cat', cat_pipe,cat_cols),
    ]
)
preprocessing_pipe
```

```python
from sklearn.linear_model import LinearRegression
model= LinearRegression()
main_pipeline= make_pipeline(preprocessing_pipe, model)
main_pipeline.fit(x_train,y_train)
```

```python
#evalation on val
from sklearn.metrics import mean_absolute_error
y_pred_val= main_pipeline.predict(x_val)
mean_absolute_error(y_val,y_pred_val)
```

```python
#test random forest 
from sklearn.tree import DecisionTreeRegressor
dt_model= make_pipeline(preprocessing_pipe,DecisionTreeRegressor())
dt_model.fit(x_train,y_train)
```

```python
y_pred_val=dt_model.predict(x_val)
mean_absolute_error(y_val,y_pred_val)
```

```python
#random forest 
from sklearn.ensemble import RandomForestRegressor
rf_model= RandomForestRegressor(n_estimators=100, random_state=42)
```

```python
rf_pipe = make_pipeline(preprocessing_pipe,rf_model)
rf_pipe.fit(x_train,y_train)
```

```python
y_pred_val=rf_pipe.predict(x_val)
mean_absolute_error(y_val,y_pred_val)
```

```python
import joblib
joblib.dump(rf_pipe,"model.pkl")
```

```python
import gradio as gr
from ultralytics import YOLO

# Load pretrained model
model = YOLO("yolov8n.pt")

# Detection function
def detect_objects(image):

    if image is None:
        return None, "Please upload an image."

    results = model.predict(image, conf=0.25)

    # Image with bounding boxes
    annotated = results[0].plot(pil=True)

    # Detected object names and confidence
    predictions = []

    for box in results[0].boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])

        name = model.names[class_id]

        predictions.append(
            f"{name}: {confidence:.2f}"
        )

    if not predictions:
        predictions = ["No objects detected."]

    return annotated, "\n".join(predictions)


# Gradio interface
demo = gr.Interface(
    fn=detect_objects,

    inputs=gr.Image(
        type="pil",
        sources=["upload", "webcam"],
        label="Upload or Capture an Image"
    ),

    outputs=[
        gr.Image(label="YOLO Detection"),
        gr.Textbox(label="Detected Objects")
    ],

    title="LingoQuest - Object Detection Test",

    description="Upload an image to test YOLO object detection."
)

demo.launch()
```
